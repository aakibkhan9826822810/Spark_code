```sql
select * from car_price
limit 10;

-- Average Price per Brand
select Brand, avg(Price) as avg_price
from car_price
group by Brand;

-- Count cars by Transmission type
select Transmission, count(*) as car_count
from car_price
group by Transmission;

-- Top 5 most expensive cars
select *
from car_price
order by Price desc
limit 5;

-- Average Mileage per Fuel_Type
select Fuel_Type, avg(Mileage) as avg_mileage
from car_price
group by Fuel_Type;

-- Count cars where Owner_Count > 1
select count(*) as owner_count_gt_1
from car_price
where Owner_Count > 1;

-- Cars with Engine_Size > 2.0
select *
from car_price
where Engine_Size > 2.0;

-- Total number of cars per Model_Year
select Model_Year, count(*) as car_count
from car_price
group by Model_Year;

-- Average Horsepower per Brand
select Brand, avg(Horsepower) as avg_horsepower
from car_price
group by Brand;

-- Cars with price greater than overall average price
select *
from car_price
where Price > (select avg(Price) from car_price);

-- Count cars grouped by Doors
select Doors, count(*) as car_count
from car_price
group by Doors;

-- Top 3 brands with highest average Price
select Brand, avg(Price) as avg_price
from car_price
group by Brand
order by avg_price desc
limit 3;

-- Second highest Price car
select *
from car_price
order by Price desc
limit 1 offset 1;

-- Cars where Mileage is above brand average
select *
from car_price c
where Mileage > (
  select avg(Mileage)
  from car_price
  where Brand = c.Brand
);

-- Rank cars within each Brand based on Price
select *, rank() over (partition by Brand order by Price desc) as price_rank
from car_price;


-- Identify duplicate cars based on (Brand, Model_Year, Mileage)
select Brand, Model_Year, Mileage, count(*) as dup_count
from car_price
group by Brand, Model_Year, Mileage
having count(*) > 1;

-- Percentage contribution of each Fuel_Type
select Fuel_Type, 
       count(*) * 100.0 / sum(count(*)) over () as fuel_type_percentage
from car_price
group by Fuel_Type;
```
