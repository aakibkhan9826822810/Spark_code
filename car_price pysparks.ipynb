```python
car = spark.read.table('data.orders.car_price')
import pyspark.sql.functions as f
from pyspark.sql.window import Window
```

```python
car.printSchema()
```

```python
# Get count of cars per Fuel_Type. 
fuel = car.groupBy('Fuel_Type').agg(f.count("Car_ID").alias('count'))
display(fuel)
```

```python
# Find average Price of all cars. 
avg = car.agg(f.round(f.avg("Price"),2).alias('avg_price'))
display(avg)
```

```python
# Find average Price per Brand. 
brand_avg = car.groupBy("Brand").agg(f.round(f.avg("Price"),2).alias('avg_price'))
display(brand_avg)
```

```python
# Count cars by Transmission type. 
tran_count = car.groupBy("Transmission").agg(f.count("Car_ID").alias('count'))
display(tran_count)
```

```python
# Find average Mileage per Fuel_Type.
fuel_avg_mileage = car.groupBy("Fuel_Type").agg(f.round(f.avg("Mileage"), 2).alias('avg'))
display(fuel_avg_mileage)
```

```python
# Count cars where Owner_Count > 1. 
owner_count = car.filter(f.col("Owner_Count") > 1).agg(f.count("Car_ID").alias("count"))
display(owner_count)
```

```python
# Get average Horsepower per Brand.
brand_horsepower = car.groupBy("Brand").agg(f.round(f.avg("Horsepower"), 2).alias('avg'))
display(brand_horsepower)
```

```python
# Find total number of cars per Model_Year. 
model_year_count = car.groupBy("Model_Year").agg(f.count("Car_ID").alias("count"))\
    .orderBy("Model_Year",ascending=False)
display(model_year_count)
```

```python
# Find cars with price greater than overall average price. 
avg_price = car.agg(f.avg('price').alias('a')).collect()[0]['a']
high_price = car.filter(f.col("Price") > avg_price)
display(high_price)
```

```python
# Find top 3 brands with highest average Price. 
brand_avg = car.groupBy("Brand").agg(f.round(f.avg("Price"),2).alias('avg_price'))\
    .orderBy('avg_price', ascending=False).limit(3)
display(brand_avg)
```

```python
# Get second highest Price car.
w = Window.orderBy(f.col("Price").desc())
second_highest = car.withColumn("rank", f.dense_rank().over(w))\
    .filter(f.col("rank") == 2)
display(second_highest)
```

```python
# Rank cars within each Brand based on Price.
w = Window.partitionBy("Brand").orderBy(f.col("Price").desc())
car_ranked = car.withColumn("rank", f.dense_rank().over(w))
display(car_ranked)
```

```python
# Get cars with highest Horsepower in each Model_Year.
w = Window.partitionBy("Model_Year").orderBy(f.col("Horsepower").desc())
car_max_hp = car.withColumn("rank", f.dense_rank().over(w)).filter(f.col("rank") == 1)
display(car_max_hp)
```
