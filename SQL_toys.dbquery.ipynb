```sql
-- 1.	Find the total number of orders.
select count(*) from orders;

-- 2.	Show all products with their launch date.
select product_name, created_at as launch_date
from products;

-- 11.	Find total revenue from all orders.
select round(sum(price_usd),2) as total_revenue
from order_items;

-- 12.	Calculate average order value (AOV).
select round(sum(price_usd)/count(distinct order_id),2) as aov
from order_items;

-- 13.	Find total orders per user.
select user_id, count(order_id) as total_orders
from orders
group by user_id;

-- 14.	Find total sessions per device type.
select device_type, count(website_session_id) as total_sessions
from website_sessions
group by device_type;

-- 15.	Calculate total revenue per product.
select product_id, round(sum(price_usd),2) as total_revenue
from order_items
group by product_id;

-- 16.	Find number of pageviews per session.
select website_session_id, count(*) as pageviews
from website_pageviews
group by website_session_id;

-- 18.	Find total sessions per marketing campaign (utm_campaign).
select utm_campaign, count(website_session_id) as total_sessions
from website_sessions
group by utm_campaign;

-- 19.	Calculate average items purchased per order.
select round(avg(items_purchased),2) as avg
from (
  select order_id, count(*) as items_purchased
  from order_items
  group by order_id
);

-- 22.	Find total revenue per product name.
select product_name, round(sum(price_usd),2) as total_revenue
from order_items o
join product_name p
```
