```sql
-- Get total orders per customer
select customer_id, count(order_id) as total_orders
from orders
group by customer_id;

-- Find total revenue per product
select p.product_id, round(sum(o.quantity * p.price),2) as rev
from products p 
join orders o
on p.product_id = o.product_id
group by p.product_id;

-- 14.	Find number of customers per city
select count(customer_id) as total, city 
from customers
group by city;


-- -- 15.	Join customers and orders to get customer names with orders
-- 16.	Join orders and products to get product names in each order
select o.order_id, c.name,p.product_name
from customers c
join orders o
on c.customer_id = o.customer_id
join products p
on o.product_id = p.product_id;


-- Get full order details: 
--   customer name 
--   product name 
--   quantity 
--   price 
select c.name as customer_name ,
  p.product_name,
  o.quantity,
  p.price,
  round(o.quantity * p.price,2) as amount
from customers c
join orders o
on c.customer_id = o.customer_id
join products p
on o.product_id = p.product_id;

--Find total spending per customer 
select c.name as customer_name, round(sum(o.quantity * p.price),2) as amount
from customers c
join orders o
on c.customer_id = o.customer_id
join products p
on o.product_id = p.product_id
group by c.name;


-- 19.	Find top 5 customers by spending
select c.name as customer_name, round(sum(o.quantity * p.price),2) as amount
from customers c
join orders o
on c.customer_id = o.customer_id
join products p
on o.product_id = p.product_id
group by c.name
order by amount desc
limit 5;

--20.	Find most sold product (by quantity)
select p.product_id, p.Product_name, sum(o.quantity) as total_sold
from products p
join orders o
on p.product_id = o.product_id
group by p.product_id, p.Product_name
order by total_sold desc
limit(3);

-- 21.	Find highest revenue generating product
select p.product_id, p.Product_name, sum(o.quantity*p.price) as total_amount
from products p
join orders o
on p.product_id = o.product_id
group by p.product_id, p.Product_name
order by total_amount desc
limit 1;

-- 22.	Find customers who never placed any order
select c.name as customer_name
from customers c
left join orders o
on c.customer_id = o.customer_id
where o.order_id is null;

-- 23.	Find products that were never ordered
select p.product_name
from products p
left join orders o
on p.product_id = o.product_id
where o.order_id is null;

-- 24.	Extract year/month from order_date
SELECT 
    order_id,
    order_date,
    date_FORMAT(order_date, 'yyyy-MM') AS year_month
FROM orders;

-- 26.	Find daily revenue
select date_format(o.order_date,'yyyy-MM-dd') as date, round(sum(o.quantity * p.price),2) as daily_rev
from products p 
join orders o
on p.product_id = o.product_id
group by date_format(o.order_date,'yyyy-MM-dd')
order by date;

-- 27.	Find customers who signed up but never ordered within 30 days 
select c.name as customer_name
from customers c
left join orders o
on c.customer_id = o.customer_id
where o.order_id is null
and datediff(current_date(), c.signup_date) < 30;


-- Rank customers by total spending
with d as (
  select c.customer_id, c.name, round(sum(o.quantity * p.price),2) as amount
  from customers c
  join orders o
  on c.customer_id = o.customer_id
  join products p
  on o.product_id = p.product_id
  group by c.customer_id,c.name )
select name, amount,
  rank() over (order by amount desc) as rank
from d;


-- 30.	Find first order date per customer
select c.customer_id,c.name, min(o.order_date) as first_order_date
from orders o
join customers c
on o.customer_id = c.customer_id
group by c.customer_id, c.name;

-- 31. Find latest product ordered per customer
with t as ( 
SELECT c.name,p.product_name,o.order_date,
ROW_NUMBER() OVER (PARTITION BY c.customer_id ORDER BY o.order_date desc) AS rn
FROM orders o
JOIN customers c
ON o.customer_id = c.customer_id
JOIN products p
ON o.product_id = p.product_id )
SELECT name,product_name,order_date
FROM t
WHERE rn = 1;



```
