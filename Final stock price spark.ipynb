```python
Add a column price_range = High - Low.
```

```python
df = spark.read.table('data.orders.stock_prices')
df = df.withColumn('price_range', df.High - df.Low)
display(df)
```

Add a column daily_return = (Close - Open) / Open.

```python
df = spark.read.table('data.orders.stock_prices')
df = df.withColumn('daily_return', (df.Close - df.Open)/df.Open).select('Date', 'daily_return')
display(df)
```
