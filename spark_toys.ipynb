```python
import pyspark.sql.functions as f
df = spark.read.table('data.default.order_items')
df = df.agg(f.round(f.sum(f.col('price_usd')), 2).alias('total'))
display(df)
```

```python
import pyspark.sql.functions as f
df = spark.read.table('data.default.order_items')
df = df.agg(
    f.countDistinct('order_id').alias('total'),
    f.sum('price_usd').alias('sum_price_usd')
)
df2 = df.withColumn('avg', f.round(f.col('sum_price_usd') / f.col('total'),2)).select('avg')
display(df2)
```

```python
df = spark.read.table('data.default.orders')
df = df.groupBy('user_id').agg(f.count('order_id').alias('total'))
display(df)
```

```python
import pyspark.sql.functions as f
df= spark.read.table('data.default.website_sessions')
df= df.groupBy('device_type').agg(f.count('website_session_id').alias('total'))
display(df)
```

```python
import pyspark.sql.functions as f
df= spark.read.table('data.default.order_items')
df= df.groupBy('product_id').agg(f.round(f.sum('price_usd'),2).alias('total'))
display(df)
```

```python
import pyspark.sql.functions as f
df= spark.read.table('data.default.website_pageviews')
df= df.groupBy('website_session_id').agg(f.round(f.count('*'),2).alias('total'))
display(df)
```

```python
import pyspark.sql.functions as f
df= spark.read.table('data.default.website_sessions')
df= df.groupBy('utm_source').agg(f.round(f.count('website_session_id'),2).alias('total'))
display(df)
```

```python
import pyspark.sql.functions as f
df = spark.read.table('data.default.order_items')
df = df.groupBy('order_id').agg(f.count('order_item_id').alias('items'))
avg_items = df.agg(f.round(f.avg('items'), 2).alias('avg'))
display(avg_items)
```
