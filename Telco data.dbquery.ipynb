```sql
-- scheck columns
SHOW COLUMNS IN telco_customer_churn;
-- check schema
DESCRIBE TABLE telco_customer_churn;

-- select sample data
SELECT * FROM telco_customer_churn LIMIT 10;

-- Count number of customers by Churn.
SELECT Churn, COUNT(*) AS customer_count
FROM telco_customer_churn
GROUP BY Churn;

-- Get all customers with tenure greater than 12 months.
SELECT *
FROM telco_customer_churn
WHERE tenure > 12;

-- Find average MonthlyCharges by Contract
SELECT Contract, AVG(MonthlyCharges) AS avg_monthly_charges
FROM telco_customer_churn
GROUP BY Contract;

-- Count customers for each InternetService type. 
SELECT InternetService, COUNT(*) AS customer_count
FROM telco_customer_churn
GROUP BY InternetService;

-- Find number of customers using both OnlineSecurity and OnlineBackup
SELECT COUNT(*) AS customer_count
FROM telco_customer_churn
WHERE OnlineSecurity = 'Yes' AND OnlineBackup = 'Yes';

-- Identify customers with SeniorCitizen = 1 and Churn = 'Yes'. 
SELECT *
FROM telco_customer_churn
WHERE SeniorCitizen = 1 AND Churn = 'Yes';

-- Find customers whose MonthlyCharges are above average
SELECT *
FROM telco_customer_churn
WHERE MonthlyCharges > (SELECT AVG(MonthlyCharges) FROM telco_customer_churn);

-- Find top 3 PaymentMethod with highest churn
SELECT PaymentMethod, COUNT(*) AS churn_count
FROM telco_customer_churn
WHERE Churn = 'Yes'
GROUP BY PaymentMethod
ORDER BY churn_count DESC
LIMIT 3;

-- Rank customers based on MonthlyCharges
SELECT *, RANK() OVER (ORDER BY MonthlyCharges DESC) AS monthlycharges_rank
FROM telco_customer_churn;

-- Find customers with duplicate customerID
SELECT customerID
FROM telco_customer_churn
GROUP BY customerID
HAVING COUNT(*) > 1;

-- Calculate churn rate by InternetService
SELECT
  InternetService,
  COUNT(*) AS total_customers,
  SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS churned_customers,
  ROUND(SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) / COUNT(*), 2) AS churn_rate
FROM telco_customer_churn
GROUP BY InternetService;

-- Identify customers with all services enabled
SELECT *
FROM telco_customer_churn
WHERE OnlineSecurity = 'Yes' AND OnlineBackup = 'Yes' AND DeviceProtection = 'Yes'
  AND TechSupport = 'Yes' AND StreamingTV = 'Yes' AND StreamingMovies = 'Yes';


/*Identify high-risk customers: 
low tenure
high MonthlyCharges
month-to-month contract */
SELECT *
FROM telco_customer_churn
WHERE tenure < 12
  AND MonthlyCharges > (SELECT AVG(MonthlyCharges) FROM telco_customer_churn)
  AND Contract = 'Month-to-month';

-- Find which Contract type contributes most to churn
SELECT Contract, COUNT(*) AS churn_count
FROM telco_customer_churn
WHERE Churn = 'Yes'
GROUP BY Contract
ORDER BY churn_count DESC
LIMIT 1;

-- Create tenure buckets (0–12, 13–24, 25+) and calculate churn count in each. 
SELECT
  CASE WHEN tenure BETWEEN 0 AND 12 THEN '0-12'
    WHEN tenure BETWEEN 13 AND 24 THEN '13-24'
    ELSE '25+'
  END AS tenure_bucket,
  COUNT(*) AS total_customers,
  SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS churn_count
FROM telco_customer_churn
GROUP BY tenure_bucket;
```
