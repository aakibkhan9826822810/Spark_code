```python
df= spark.read.csv('/Volumes/data/orders/files/archive/customers.csv', header=True,inferSchema=True)
import pyspark.sql.functions as f
```

first check the sample data and schema 

```python
df.printSchema()
display(df.limit(10))
```

check duplicates

```python
import pyspark.sql.functions as f
dup_records =df.groupBy('customer_id').agg(f.count('*').alias('count')).where(f.col('count')>1)
display(dup_records.count())
```

check nulls


```python
null_records = df.select([
    f.count(f.when(f.col(c).isNull(), c)).alias(c)
    for c in df.columns
])
display(null_records)
```

check word fomations in gender

```python
word_formation = df.select('gender').distinct()
display(word_formation)
```

check age should be correct 

```python
curr_age = df.where(f.col('age')<=0).count()
display(curr_age)
```

get quick summary of table 

```python
df.describe().show()
```

check data correctness

```python
date_curr = df.filter(f.col("join_date")>f.current_date()).count()
display(date_curr)
```

now save data in delta format

```python
df.write.mode("overwrite").format('delta').parquet('/Volumes/data/orders/files/chocolate/customers')
```
