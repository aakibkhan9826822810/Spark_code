```python
df= spark.read.csv('/Volumes/data/orders/files/archive/products.csv', header=True,inferSchema=True)
import pyspark.sql.functions as f
```

first check the sample data and schema 

```python
df.printSchema()
display(df.limit(10))
```

check duplicates

```python
import pyspark.sql.functions as f
dup_records =df.groupBy('product_id').agg(f.count('*').alias('count')).where(f.col('count')>1)
display(dup_records.count())
```

check nulls


```python
null_records = df.select([
    f.count(f.when(f.col(c).isNull(), c)).alias(c)
    for c in df.columns
])
display(null_records)
```

check word fomations

```python
# for brand
word_formation = df.select('brand').distinct()
display(word_formation)
```

```python
# for category
word_formation = df.select('category').distinct()
display(word_formation)
```

use decribe for quiek summary

```python
df.describe().show()
```

check nemeric validation


```python
per = df.where((f.col("cocoa_percent")<0) | (f.col("cocoa_percent")>100)).count()
display(per)
weight = df.where(f.col("weight_g")<=0).count()
display(weight)
```
