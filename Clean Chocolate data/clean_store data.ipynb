```python
df = spark.read.csv('/Volumes/data/orders/files/archive/stores.csv', header=True,inferSchema=True)
```

first check the sample data and schema 


```python
df.printSchema()
display(df.limit(10))
```

check duplicates


```python
import pyspark.sql.functions as f
dup_records =df.groupBy('store_id').agg(f.count('*').alias('count')).where(f.col('count')>1)
display(dup_records.count())
```

check if one store id is mape with multiple store

```python
store_dup= df.groupBy("store_id")\
    .agg(f.countDistinct("store_name").alias("name_count")) \
    .filter("name_count > 1")
display(store_dup)
```

check nulls

```python

null_records = df.select([
    f.count(f.when(f.col(c).isNull(), c)).alias(c)
    for c in df.columns
])
display(null_records)
```

check work formations

```python
#for store type
store_type_formation = df.select('store_type').distinct()
display(store_type_formation)
```

```python
#for store type
city_formation = df.select('city').distinct()
display(city_formation)
```

```python
# for country
country_formation = df.select('country').distinct()
display(country_formation)
```

save it in detla format


```python
df.write.mode("overwrite").format('delta').parquet('/Volumes/data/orders/files/chocolate/stores')
```
