first create a data frame 


```python
data = [
    (1101,'Aakib',56,60,70,56,74),
    (1102,'Aatif',78,62,86,76,84),
    (1103,'Azaz',76,68,76,86,70),
    (1104,'Ayaz',75,80,90,66,78),
    (1105,'Faiyaz',78,95,80,67,89),
]
columns =['roll_number','name','maths','physics','chemistry','english','hindi']

marks = spark.createDataFrame(data,columns)
display(marks)
```

write in delta format


```python
marks.write.mode('overwrite').format('delta').save('/Volumes/data/orders/files/Schemas/marks')
```

create new data frame 

```python
data2 = [
    (1106,'Muskan',None,70,70,89,78,90),
    (1107,'sohel',78,95,86,76,None,89),
    (1108,'javed',66,67,76,None,89,75),
    (1109,'asif',73,80,90,89,None,65),
    (1110,'imran',89,92,80,None,87,98),
    (1112,'anas',None,70,70,42,56,89),
    (1113,'mohsin',None,70,45,89,48,70),
    (1114,'moin',None,70,77,89,78,85),
]
columns2 =['roll_number','name','maths','physics','chemistry','english','hindi','biology']

marks2 = spark.createDataFrame(data2,columns2)
display(marks2)
```

again right it into same folder

```python
marks2.write.mode('append').option('mergeSchema','true').format('delta').save('/Volumes/data/orders/files/Schemas/marks')
```

now read data

```python
result = spark.read.format('delta').load('/Volumes/data/orders/files/Schemas/marks')
display(result)
```

used same process in parquet


```python
marks.write.mode('append').format('parquet').save('/Volumes/data/orders/files/Schemas/parquet')
marks2.write.mode('append').format('parquet').save('/Volumes/data/orders/files/Schemas/parquet')
result = spark.read.parquet('/Volumes/data/orders/files/Schemas/parquet')
display(result)
```

```python
from delta.tables import DeltaTable
df = DeltaTable.forPath(spark, "/Volumes/data/orders/files/Schemas/marks").history()
display(df)
```
