Repartition

```python
df=spark.read.format("delta").load("/Volumes/data/orders/files/chocolate/stores/")
display(df)
```

check current number of partitions

```python
# check current number of partitions
from pyspark.sql.functions import spark_partition_id
df.select(spark_partition_id()).distinct().count()
```

check the total rows per partition

```python
# check the total rows per partition
from pyspark.sql.functions import spark_partition_id
df.groupBy(spark_partition_id()).count().display()
```

repartiton it then check the total number of rows


```python
# repartiton it then check the total number of rows
par = df.repartition(5)
par.groupBy(spark_partition_id()).count().display()
```

repartion by column

```python
# repartion by column
city=spark.read.format('delta').load('/Volumes/data/orders/files/chocolate/sales/').repartition(5,'customer_id')
city.groupBy(spark_partition_id()).count().display()
```

colease

with colease decrese the number of partions

```python
col = par.coalesce(3)
col.groupBy(spark_partition_id()).count().display()
```
