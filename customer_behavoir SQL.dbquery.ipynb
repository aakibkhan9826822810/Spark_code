```sql
select * from customer_shopping_behavior
limit 10;

-- Count number of customers by Gender. 
select count('customer_id') as count, gender 
from customer_shopping_behavior 
group by gender;

-- Find total purchase amount by Category. 
select category, round(sum(`Purchase Amount (USD)`), 2) as total_purchase_amount
from customer_shopping_behavior
group by category;

-- Get average purchase amount by Gender. 
select gender, round(avg(`Purchase Amount (USD)`), 2) as amount
from customer_shopping_behavior
group by Gender;

-- Count customers by Subscription Status. 
select `Subscription Status`, count('customer_id') as count
from customer_shopping_behavior
group by `Subscription Status`;

-- Find top 5 locations with highest total spending. 
select location,round(sum(`purchase amount (usd)`),2) as amount
from customer_shopping_behavior
group by location
order by amount desc
limit 5;

-- Get number of purchases by Season. 
select Season, count('customer_id') as count
from customer_shopping_behavior
group by Season;

-- Find most commonly used Payment Method. 
select `payment method`, count('customer_id') as count
from customer_shopping_behavior
group by `payment method`
order by count desc
limit 1;

-- Get average Review Rating per category. 
select Category, round(avg(`Review Rating`),2) as rating
from customer_shopping_behavior
group by Category;

-- Count how many times discount was applied. 
select count(*) from customer_shopping_behavior
where `discount applied` = 'Yes';

-- Get total revenue generated by each Shipping Type
select `shipping type`,round(sum(`purchase amount (usd)`),2) as amount
from customer_shopping_behavior
group by `shipping type`;


-- Rank customers based on total purchase amount. 
with amount as (
  select `customer id`, round(sum(`purchase amount (usd)`),2) as amount
  from customer_shopping_behavior
  group by `customer id`
)
select *,rank() over (order by amount desc) as rank
from amount;

-- Find top 3 customers per Location by spending. 
with amount as (
  select `customer id`, location, round(sum(`purchase amount (usd)`),2) as amount
  from customer_shopping_behavior
  group by `customer id`, location
),
rank as (
  select *,rank() over(partition by location order by amount desc) as rank
from amount)
select * from rank 
where rank <= 3;

-- Calculate percentage contribution of each Category to total sales. 
select category, round(sum(`purchase amount (usd)`)*100/(select sum(`purchase amount (usd)`) from customer_shopping_behavior)
          ,2) as percent
from customer_shopping_behavior
group by category;

-- Find customers whose purchase amount is above overall average.
select *
from customer_shopping_behavior
where `Purchase Amount (USD)` > (select avg(`Purchase Amount (USD)`) from customer_shopping_behavior);

-- Identify repeat customers (based on Previous Purchases).
select distinct `Customer ID`
from customer_shopping_behavior
where `Previous Purchases` > 1;

-- Find most popular item in each Category. 
with top_cat as (
  select Category, `Item Purchased`, count(`customer id`) as count
  from customer_shopping_behavior
  group by Category, `Item Purchased`
),
rank as (
  select *,rank() over(partition by Category order by count desc) as rank
from top_cat)
select Category, `Item Purchased` from rank 
where rank =1;
```
