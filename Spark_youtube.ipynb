```python
import  pyspark.sql.functions as f
df = spark.read.table('data.orders.youtube')

df = df.groupBy('channel_name').agg(f.sum('Likes').alias('otal_likes'), f.sum('comments').alias('total_comments'))
display(df)
```

```python
import pyspark.sql.functions as f
df = spark.read.table('data.orders.youtube')
df = df.select('Views_Per_Day','video_id','title').orderBy(f.col('Views_Per_Day').desc()).limit(5)
display(df)
```

```python
import pyspark.sql.functions as f
from pyspark.sql.window import Window
df = spark.read.table('data.orders.youtube')
w = Window.orderBy(f.col('Views_Per_Day').desc())
df = df.withColumn('rk', f.row_number().over(w))\
    .withColumn('drk', f.dense_rank().over(w)).select('video_id','title','Views_Per_Day','rk','drk')
display(df)
```

```python
import pyspark.sql.functions as f
df = spark.read.table('data.orders.youtube')

df = df.withColumn(
    'Video_Cat',
    f.when(f.col('Views') > 1_000_000, 'Viral').otherwise('Normal')
)
display(df)
```

```python
import pyspark.sql.functions as f
from pyspark.sql.window import Window
df = spark.read.table('data.orders.youtube')
w = Window.orderBy('video_id').rowsBetween(-2, 0)
df = df.withColumn('rolling_avg_views',f.round(f.avg('Views').over(w),2))\
    .select('Video_ID','Title','Views','rolling_avg_views')
display(df)
```

```python
import pyspark.sql.functions as f
from pyspark.sql.window import Window
df = spark.read.table('data.orders.youtube')
w = Window.orderBy(f.col('views').desc())
df = df.withColumn('rk',f.rank().over(w))\
    .select('Video_ID','Views','rk')
display(df)
```
